```python
import matplotlib.pyplot as plt
import numpy as np
```

## Modello di crescita discreto di Ricker

```python
def ComputeRickerModel(x0, r, k, iterations):
    x = x0
    y = [x]
    for i in range(iterations):
        new_x = x*np.exp(r*(1-x/k))
        x = new_x
        y.append(x)
    return y

def ComputeRiggerModel(x0, r, k, iterations):
    x = x0
    y = [x]
    for i in range(iterations):
        new_x = x + (np.exp(r*(1-x/k)) - x)
        x = new_x
        y.append(x)
    return y
```

```python
iterations = 1000
r0 = 2
k = 300
x0 = 10


y = ComputeRickerModel(x0, r0, k, iterations)
z = ComputeRiggerModel(x0, r0, k, iterations)

#plot the range 0 to 10
plt.plot(y)
plt.plot(z)
```

```python
r_scale = np.linspace(0, 4, 1000)
asymptotic = [ComputeRickerModel(x0, r_scale, k, iterations)[-1] for r_scale in r_scale]
plt.scatter(r_scale, asymptotic)
plt.title('Asymptotic value of the Ricker model')
plt.xlabel('r')
plt.ylabel('asymptotic value')
plt.show()
```
