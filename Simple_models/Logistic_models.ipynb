```python
import matplotlib.pyplot as plt
import numpy as np
```

## Logistic Model



```python
def LogisticMap(r0, k, x):
    return r0*x*(1-x/k)

def ComputeLogisticModel(x0, r0, k, iterations):
    x = x0
    y = [x]
    for i in range(iterations):
        new_x = x + LogisticMap(r0, k, x)
        x = new_x
        y.append(x)
    return y
```

```python
iterations = 1000
r0 = 0.01
k = 300
x0_top = 400
x0_bottom = 10
delta_t = 0.01

y_top = ComputeLogisticModel(x0_top, r0, k, iterations)
y_bottom = ComputeLogisticModel(x0_bottom, r0, k, iterations)

plt.plot(y_top, label='x0 = 1000')
plt.plot(y_bottom, label='x0 = 10')
plt.plot([k]*iterations, label='carrying capacity')
plt.legend()
plt.show()
```

Provando a perturbare il sistema in un suo punto di equilibrio

```python
def LogisticMap(r0, k, x):
    return r0*x*(1-x/k)

def ComputeLogisticModel_perturbed(x0, r0, k, iterations, perturbation_t):
    x = x0
    y = [x]
    for i in range(iterations):
        if i == perturbation_t:
            x = x + 50
        new_x = x + LogisticMap(r0, k, x)
        x = new_x
        y.append(x)
    return y
```

```python
iterations = 1000
r0 = 0.05
k = 300
x0 = 100
perturbation_t = 500

y = ComputeLogisticModel_perturbed(x0, r0, k, iterations, perturbation_t)

plt.plot(y, label='x0 = 100')
```

Si noti come dopo la perturbazione il sistema ritorna al suo stato precedente con velocità -r0. Quindi maggiore è r0 più velocemente ritornerà al suo stato di equilibrio

```python
r_scale = np.linspace(0, 4, 1000)
asymptotic = [ComputeLogisticModel(x0, r_scale, k, iterations)[-1] for r_scale in r_scale]
plt.scatter(r_scale, asymptotic)
plt.title('Asymptotic value of the logistic model')
plt.xlabel('r')
plt.ylabel('asymptotic value')
plt.show()
```
